## 3. Read the website's CSV

The cell below loads the actual CSV used by the website. `index_col="location_id"` uses each resource's identifier as its row label, so we can select a row with `.loc["HB03"]`.

The columns describe the resource's name, category, area, access information, and official webpage. Find Morrison Library in the displayed table and compare it with the website.

```python
import pandas as pd

csv_path = "docs/_data/locations.csv"
locations = pd.read_csv(csv_path, index_col="location_id")
locations
```

## 4. Improve one entry and add a resource

Open an existing resource's official webpage. Improve its access note with a specific, useful detail you can verify there. For example, clarify who can use it or where to check visiting arrangements. Replace the two placeholder strings below; keep the identifier of a resource already in the table.

`.loc[row_label, column_name]` selects one cell. Assigning a string changes that cell in the DataFrame. The last line displays your updated row. This has not saved the CSV yet.

```python
existing_id = "HB03"  # Editing the Morrison Library.
locations.loc[existing_id, "access_note"] = "Morrison is generally open between 10:00AM and 5:00PM on weekdays. Please check the official website for the most up to date information." 
locations.loc[existing_id, "source_url"] = "https://www.lib.berkeley.edu/visit/morrison"
locations.loc[existing_id]
```

Now choose one Berkeley place or service that is not already listed. Check its official webpage and replace every placeholder in this dictionary. Use public information and retain all eight original resources.

The dictionary keys match the CSV columns. `.loc["HB09"] = new_resource` puts those values in a row labeled `HB09`. Running it again replaces that same row, so rerunning the notebook does not add duplicates.

```python
new_resource = {
    "name": "UC Berkeley Career Center",
    "category": "student support",
    "area": "UC Berkeley",
    "access_note": "The Career Center offers career planning essentials, events and workshops, career counseling appointments, job and internship search resources and support, and planning for graduate and professional school to undergraduates, graduate students, and alumni.",
    "source_url": "https://career.berkeley.edu/",
}

locations.loc["HB09"] = new_resource
locations
```

### Save your data changes

Check the table above before saving: nine rows, all eight original identifiers (`HB01`–`HB08`), your improved entry, and your new `HB09` entry. Replace every placeholder and make sure both source links open the official pages you checked.

The first line below writes the DataFrame back to the website's CSV, including the `location_id` row labels. The second line reads the saved file again so you can verify what is on disk. Changing a DataFrame in memory does not update the file until you call `to_csv`.
